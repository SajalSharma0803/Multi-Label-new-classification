```python
# Necessary Imports
import torch
import torch.nn.functional as F
import torchvision.transforms as T
import timm
from PIL import Image

# Configuration used during training (must match your YAML)
DEVICE = 'cuda' if torch.cuda.is_available() else 'cpu'
STUDENT_MODEL_NAME = 'resnet18'
NUM_CLASSES = 10
IMG_SIZE = 224
CHECKPOINT_PATH = './outputs/10class/student_epoch40.pt'
```

```python
def build_transforms(img_size):
    train_t = T.Compose([
        T.RandomResizedCrop(img_size, scale=(0.2,1.0)),
        T.RandomHorizontalFlip(),
        T.ToTensor(),
        T.Normalize(mean=[0.485,0.456,0.406], std=[0.229,0.224,0.225]),
    ])
    val_t = T.Compose([
        T.Resize(int(img_size*1.15)),
        T.CenterCrop(img_size),
        T.ToTensor(),
        T.Normalize(mean=[0.485,0.456,0.406], std=[0.229,0.224,0.225]),
    ])
    return train_t, val_t
```

```python
# 1. Instantiate the student model
student = timm.create_model(STUDENT_MODEL_NAME, pretrained=False, num_classes=NUM_CLASSES)
student.to(DEVICE)

# 2. Load the checkpoint
checkpoint = torch.load(CHECKPOINT_PATH, map_location=DEVICE)

# Note: We only need the student's weights ('student_state') for prediction
student.load_state_dict(checkpoint['student_state'])

# 3. Set to evaluation mode! This disables dropout, batchnorm updates, etc.
student.eval()

print("Model loaded and set to evaluation mode.")
```

```python
# --- CHANGE THIS TO YOUR IMAGE FILE PATH ---
single_image_path = '/content/accordian (19).jpg'

# 1. Load Image
try:
    img = Image.open(single_image_path).convert('RGB')
except FileNotFoundError:
    print(f"Error: Image file not found at {single_image_path}")
    # return or exit here if running a script

# 2. Apply Validation Transforms
_, val_transform = build_transforms(IMG_SIZE)
img_tensor = val_transform(img)

# 3. Add a batch dimension (B=1) and move to device
img_tensor = img_tensor.unsqueeze(0).to(DEVICE)

# 4. Run Inference
with torch.no_grad():
    # Forward pass outputs the logits (raw scores)
    logits = student(img_tensor)

# 5. Convert Logits to Probabilities and Prediction
probabilities = F.softmax(logits, dim=1)

# Find the class with the highest probability
confidences, predicted_index = torch.max(probabilities, 1)

# Get the final results
predicted_class = predicted_index.item()
confidence = confidences.item() * 100

print("\n--- Prediction Results ---")
print(f"Predicted Class Index: {predicted_class}")
print(f"Confidence: {confidence:.2f}%")
# Optional: Print the full probability distribution
# print("Full Probabilities:", probabilities.squeeze(0).cpu().numpy())
```
